```python
all_fighters_filtered = all_fighters[all_fighters['total_fights'] >= 4]
all_fighters_filtered.info()
```

```python
all_fighters_filtered['Z-Score'] = (all_fighters_filtered['sub_win_ratio'] - all_fighters_filtered['sub_win_ratio'].mean()) / all_fighters_filtered['sub_win_ratio'].std()
all_fighters_filtered = all_fighters_filtered.sort_values(by='Z-Score', ascending=False)
all_fighters_filtered.head(20)
```

```python
all_fighters_filtered.to_csv('fighters_z_scores.csv', index=False)
```
