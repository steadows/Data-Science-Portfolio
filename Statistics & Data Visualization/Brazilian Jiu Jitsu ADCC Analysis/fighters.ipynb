```python
from geopy.geocoders import Nominatim
from geopy.exc import GeocoderTimedOut
import time

# Initialize the geolocator
geolocator = Nominatim(user_agent="geoapiExercises")

location_cache = {}

def get_lat_long_cached(row):
    location_key = f"{row['city']}, {row['country']}"
    if location_key in location_cache:
        return location_cache[location_key]
    try:
        location = geolocator.geocode(location_key)
        if location:
            location_cache[location_key] = (location.latitude, location.longitude)
            return location.latitude, location.longitude
    except Exception as e:
        print(f"Error: {e}")
    location_cache[location_key] = (None, None)
    return None, None

world_fighters[['latitude', 'longitude']] = world_fighters.apply(
    lambda row: get_lat_long_cached(row), axis=1, result_type="expand"
)

```
